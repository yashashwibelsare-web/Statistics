```python
# 1.Smartphone Screen Time
```

```python
s = [9.8, 10.3, 10.1, 9.9, 10.4,10.2, 9.7, 10.0, 10.5, 9.8,10.1, 9.9, 10.2, 10.3, 10.0,9.8, 10.4, 10.1, 9.9, 10.2]
```

```python
mn = np.mean(s)
mn
```

```python
sd = np.std(s)
sd
```

```python
n = 20
dof = n-1
dof
```

```python
from scipy.stats import t
```

```python
tscore = t.ppf(0.95,dof)
tscore
```

```python
moe = tscore * (sd/np.sqrt(n))
moe
```

```python
ci = (mn-moe,mn+moe)
ci
```
